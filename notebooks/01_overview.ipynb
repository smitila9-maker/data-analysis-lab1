```python
import pandas as pd

# Читаем наш отладочный срез!
file_path = '../data/interim/sample_200k.csv'
df = pd.read_csv(file_path)

# 1. Число строк и столбцов
print(f"Размер таблицы: {df.shape[0]} строк, {df.shape[1]} столбцов\n")

# 2. Типы данных, число уникальных значений и доля пропусков
info_df = pd.DataFrame({
    'Тип данных': df.dtypes,
    'Уникальных значений': df.nunique(),
    'Доля пропусков (%)': (df.isna().sum() / len(df)) * 100
})

# Выводим красивую табличку
info_df
```
