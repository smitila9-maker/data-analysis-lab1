4. Анализ пропусков

```python
import pandas as pd

# Читаем файл. В Colab он лежит просто в корневой папке
df = pd.read_csv('sample_200k.csv')

# 1. Число строк и столбцов
print(f"Размер таблицы: {df.shape[0]} строк, {df.shape[1]} столбцов\n")

# 2. Перечень столбцов с типами и долей пропусков
info_df = pd.DataFrame({
    'Тип данных': df.dtypes,
    'Доля пропусков (%)': (df.isna().sum() / len(df)) * 100
})
display(info_df) # В Colab лучше использовать display вместо print для красивых таблиц
```

5. Визуализация

```python
import matplotlib.pyplot as plt

# Настройка размера графика
plt.figure(figsize=(10, 5))

# Строим гистограмму по столбцу altitude_1
df['altitude_1'].hist(bins=50, color='skyblue', edgecolor='black')

# Подписываем оси и заголовок
plt.title('Распределение высоты (altitude_1)')
plt.xlabel('Высота (футы)')
plt.ylabel('Количество рейсов')
plt.show()
```
